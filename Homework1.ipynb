```python
import math
a,b,c=map(float,input('nhap 3 canh cua tam giac').split())
if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print(f'Dien tich tam giac la: {s:.1f}')
else:
    print("Khong phai 3 canh cua tam giac")
```

```python
ten = input("Ten chu ho: ")
cu = int(input("Chi so thang truoc: "))
moi = int(input("Chi so thang nay: "))
a=moi - cu
tien_dien = 0
if a > 400:
    tien_dien += (a - 400) * 3460
    a = 400
if a > 300:
    tien_dien += (a - 300) * 3350
    a = 300
if a > 200:
    tien_dien += (a - 200) * 2998
    a = 200
if a > 100:
    tien_dien += (a - 100) * 2380
    a = 100
if a > 50:
    tien_dien += (a - 50) * 2050
    a = 50
if a > 0:
    tien_dien += a * 1984
tong_tien=tien_dien*1.08
print(f"Ho va ten: {ten}")
print(f"Tien phai tra la: {tong_tien:.0f}")
```
